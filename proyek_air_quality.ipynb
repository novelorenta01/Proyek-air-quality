## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Dari 12 stasiun pemantauan di Beijing, stasiun mana yang udara-nya paling bersih dan stasiun mana yang paling kotor selama periode 2013–2017?

  SMART-nya:
    - Specific: perbandingan 12 stasiun di beijing
    - Measurable: nilai dari PM2.5, 'PM10', 'SO2', 'NO2', 'CO', 'O3
    - Action-oriented: dapat menjadi dasar untuk mengevaluasi udara yang bersih dan kotor
    - Relevant: berkaitan lansung dengan stasiun pemantuan
    - Time-bound: tahun 2013-2017

- **Pertanyaan 2:** Bagaimana fluktuasi tingkat polusi udara lintas musim (Semi, Panas, Gugur, Dingin) di Beijing sepanjang tahun 2013–2017, dan polutan gas apa yang paling mendominasi lonjakan polusi di musim terburuk?

  SMART-nya:
  - Specific: analisis perubahan musiman (spring, summer, autumn, winter)
  - Measurable: angka rata-rata konsentrasi polutan antar musim
  - Action-oriented: Membantu  merancang regulasi berbahan bakar batu bara menjelang musim dingin.
  - Relevant: berkaitan dengan polusi udara dan lintas musim
  - Time-bound: tahun 2013-2017

- **Pertanyaan 3:** Seberapa besar pengaruh curah hujan (RAIN) dan kecepatan angin (WSPM) terhadap penurunan konsentrasi berbagai polutan udara di sepanjang rentang waktu 2013–2017?

  SMART-nya:
  - Specific: pengaruh curah hujan dan kecepatan angin.
  - Measurable: nilai koefisien korelasi Pearson ($r$) pada matriks statistik (-1 hingga 1).
  - Action-oriented: Memberikan acuan bagi sistem peringatan dini publik berdasarkan prediksi kecepatan angin untuk mitigasi polusi.
  - Relevant: relevan untuk memahami distribusi pasar berdasarkan lokasi.
  - Time-bound: tahun 2013-2017

- **Pertanyaan 4:** Berapa besar akumulasi frekuensi jam paparan masyarakat terhadap berbagai kategori kualitas udara (PM2.5) di seluruh wilayah Beijing selama periode 2013–2017?

  SMART-nya:
  - Specific: pengelompokan tingkat keparahan PM2.5 ke dalam kategori status udara (Baik, Sedang, Tidak Sehat, Berbahaya) menggunakan teknik binning.
  - Measurable: jumlah total jam kejadian dan persentase waktu pada setiap kategori kualitas udara.
  - Action-oriented: Menjadi dasar evaluasi kesehatan masyarakat untuk memperketat pembatasan aktivitas luar ruangan saat kategori udara masuk level "Berbahaya".
  - Relevant: krusial untuk menerjemahkan angka  mentah menjadi informasi standar kesehatan yang mudah dipahami publik
  - Time-bound: tahun 2013-2017

**Pertanyaan 2**:

```python
# visual no 2
# kelompokkan data berdasarkan musim dan menghitung rata-ratanya
polutan_cols = ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3']
df_season = all_data_df.groupby('season')[polutan_cols].mean()

urutan_musim = ['Musim semi (spring)', 'Musim panas (summer)', 'Musim gugur (autumn)', 'Musim dingin (winter)']
df_season = df_season.reindex(urutan_musim)

# plot chart
fig, ax1 = plt.subplots(figsize=(10, 6))

# Sumbu Y Pertama (Kiri): Untuk CO (karena nilainya sangat besar/ribuan)
color = 'tab:red'
ax1.set_xlabel('Musim', fontsize=11, fontweight='bold')
ax1.set_ylabel('Konsentrasi CO ($\mu g/m^3$)', color=color, fontsize=11, fontweight='bold')
line1 = ax1.plot(df_season.index, df_season['CO'], color=color, marker='o', linewidth=2.5, label='CO (Sumbu Kiri)')
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(axis='x', linestyle='--', alpha=0.5)

# Sumbu Y Kedua (Kanan): Untuk polutan lain (PM2.5, PM10, SO2, NO2, O3)
ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Konsentrasi Polutan Lain ($\mu g/m^3$)', color='black', fontsize=11, fontweight='bold')

# Plot polutan lainnya
lines_other = []
for col in ['PM2.5', 'PM10', 'SO2', 'NO2', 'O3']:
    l = ax2.plot(df_season.index, df_season[col], marker='o', linewidth=1.5, label=col)
    lines_other.append(l[0])

ax2.tick_params(axis='y')

# Judul dan Legenda
plt.title('Fluktuasi Polusi Udara Lintas Musim di Beijing (2013-2017)\n(Dominasi Lonjakan Gas CO di Musim Dingin)', fontsize=13, fontweight='bold', pad=15)

# Menggabungkan legenda dari kedua sumbu
lines = line1 + lines_other
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper center', bbox_to_anchor=(0.5, -0.18), ncol=3, frameon=True)

plt.tight_layout()
plt.show()
```

## Analisis Lanjutan (Opsional)

```python
# clustering analisis lanjutan
# binning untuk PM2.5 (Kualitas Partikulat Halus)
bins_pm25 = [0, 35, 75, 150, 500]
labels_pm25 = ['Baik', 'Sedang', 'Tidak Sehat', 'Berbahaya']
all_data_df['Cat_PM2.5'] = pd.cut(all_data_df['PM2.5'], bins=bins_pm25, labels=labels_pm25)
print(all_data_df['Cat_PM2.5'].value_counts())
frekuensi_pm25 = all_data_df['Cat_PM2.5'].value_counts()
```

```python
# visual binning PM2.5
plt.figure(figsize=(10, 5))
ax = frekuensi_pm25.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Frekuensi Kualitas Udara Berdasarkan Kategori PM2.5 (2013-2017)', fontsize=12, fontweight='bold', pad=15)
plt.xlabel('Kategori Kualitas Udara (PM2.5)', fontsize=10, fontweight='bold')
plt.ylabel('Jumlah Kejadian (Jam)', fontsize=10, fontweight='bold')
plt.xticks(rotation=30, ha='right')

# menambahkan angka di atas batang grafik agar informatif
for p in ax.patches:
    ax.annotate(format(p.get_height(), ','),
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center',
                xytext=(0, 5),
                textcoords='offset points',
                fontsize=9)

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Kategori "Baik" mencatatkan durasi waktu terbanyak, yaitu mencapai 156.965 jam. Secara kumulatif dalam 5 tahun, sebagian besar waktu di Beijing masih didominasi oleh udara bersih atau kategori aman.
- Akumulasi Bahaya yang Signifikan Jika digabungkan, kategori Tidak Sehat (100.528 jam), Sedang (99.415 jam), dan Berbahaya (62.911 jam) memakan waktu total ratusan ribu jam. Warga Beijing menghabiskan proporsi waktu yang sangat besar terpapar udara tercemar, yang menunjukkan bahwa polusi udara merupakan masalah kronis dan sistemik, bukan sekadar kejadian sesaat.
